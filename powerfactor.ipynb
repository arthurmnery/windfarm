```python
import boto3
import json
from random import uniform
import time
from datetime import datetime


cliente = boto3.client('kinesis',aws_access_key_id='ACCESS_KEY', aws_secret_access_key='SECRET_ACCESS_KEY', region_name='REGION_NAME')

id = 0
while True :
  dados = uniform(0.7,1)
  id += 1;
  registro = {'idtemp' : str(id), 'data' : str(dados), 'type' : 'powerfactor', 'timestamp' : str(datetime.now())}
  cliente.put_record(StreamName = 'STREAMNAME', Data = json.dumps(registro), PartitionKey='02')
  time.sleep(10)
```
